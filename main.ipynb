```python
import pandas as pd
from pathlib import Path

input_file = Path("SET50_aligned.csv")
output_base_dir = Path(r"D:\SET50_direction_prediction_paper\data-folds")
output_base_dir.mkdir(parents=True, exist_ok=True)

df = pd.read_csv(input_file, index_col='Date', parse_dates=True)

test_years = [2022, 2023, 2024, 2025]
start_year = 2012

print("--- GENERATING WALK-FORWARD FOLDS ---")

for i, test_year in enumerate(test_years, start=1):
    fold_dir = output_base_dir / f"fold_{i}"
    fold_dir.mkdir(parents=True, exist_ok=True)
    
    train_end_year = test_year - 1
    
    train_df = df[(df.index.year >= start_year) & (df.index.year <= train_end_year)]
    test_df = df[df.index.year == test_year]
    
    train_filename = fold_dir / f"train_{start_year}_{train_end_year}.csv"
    test_filename = fold_dir / f"test_{test_year}.csv"
    
    train_df.to_csv(train_filename)
    test_df.to_csv(test_filename)
    
    print(f"Fold {i} Created:")
    print(f"  Train: {start_year}-{train_end_year} | Rows: {len(train_df)} | File: {train_filename.name}")
    print(f"  Test:  {test_year} | Rows: {len(test_df)} | File: {test_filename.name}")
```
