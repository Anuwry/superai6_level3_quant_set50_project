# Ridge: Actual vs Predicted
กราฟนี้ใช้เฉพาะ saved out-of-sample predictions จาก test folds และไม่ train โมเดลซ้ำ

```python
from pathlib import Path
import sys

PROJECT_ROOT = next(path for path in (Path.cwd(), *Path.cwd().parents) if (path / 'models').is_dir() and (path / 'outputs').is_dir())
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from IPython.display import display
import matplotlib.pyplot as plt
from models.prediction_visualization import (discover_experiments, find_project_root, load_predictions, metrics_by_fold, plot_predictions, regression_metrics)

MODEL_ALIASES = ('ridge',)
EXPERIMENT_NAME = None  # เช่น 'ridge_regression_full_non_ta'; None = แสดงทุกชุดที่พบ
```

```python
experiments = discover_experiments(PROJECT_ROOT, MODEL_ALIASES)
if EXPERIMENT_NAME is not None:
    experiments = [path for path in experiments if path.name == EXPERIMENT_NAME]
if not experiments:
    raise FileNotFoundError(f'No Ridge predictions found for {EXPERIMENT_NAME=}')
display([str(path.relative_to(PROJECT_ROOT)) for path in experiments])
```

```python
for experiment in experiments:
    predictions = load_predictions(experiment)
    print(f'\n{experiment.relative_to(PROJECT_ROOT)} | rows={len(predictions):,}')
    display(metrics_by_fold(predictions).style.format('{:.4f}'))
    display(regression_metrics(predictions))
    plot_predictions(predictions, experiment.name)
    plt.show()
```
